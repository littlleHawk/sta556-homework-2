```python
FILE_PATH = "../data/raw/thalanga.geoh5"
```

```python
hf_df = h5py.File(FILE_PATH, "r")
hf_df.keys()
```

```python
data_sets = hf_df["GEOSCIENCE"]
print(data_sets.keys())
```

```python
keys = [key for key in data_sets["Groups"]]
data_sets["Groups"][keys[0]]
```

```python
# List datasets stored in the geoh5 file
datasets = []

def collect_datasets(name, obj):
    if isinstance(obj, h5py.Dataset):
        datasets.append((name, obj.shape, obj.dtype))

hf_df.visititems(collect_datasets)
data_specs = pd.DataFrame(datasets, columns=["path", "shape", "dtype"])

data_specs.head()
```
